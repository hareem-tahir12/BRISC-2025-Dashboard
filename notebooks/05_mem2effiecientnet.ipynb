```python
import os

dataset_dir = "/content/drive/MyDrive/BRISC_ML_Project/classification_task"

train_dir = os.path.join(dataset_dir, "train")
test_dir = os.path.join(dataset_dir, "test")

print("Train path:", train_dir)
print("Test path:", test_dir)
```

```python


IMG_SIZE = (600, 600)
BATCH_SIZE = 4

train_ds = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=42,
    validation_split=0.20,
    subset="training"
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False,
    seed=42,
    validation_split=0.20,
    subset="validation"
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("Classes:", train_ds.class_names)
```

```python
from tensorflow.keras.applications import EfficientNetB7
from tensorflow.keras import layers, models

base_model = EfficientNetB7(
    include_top=False,
    weights="imagenet",
    input_shape=(600, 600, 3)
)

base_model.trainable = False

model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(4, activation="softmax")
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully!")
```

```python
from tensorflow.keras.callbacks import ModelCheckpoint

checkpoint = ModelCheckpoint(
    "/content/drive/MyDrive/BRISC_ML_Project/features/efficientnetb7_best.keras",
    monitor="val_accuracy",
    save_best_only=True,
    verbose=1
)
```

```python
from PIL import Image
import os

bad_files = []

for root, dirs, files in os.walk(train_dir):
    for file in files:
        path = os.path.join(root, file)

        try:
            with Image.open(path) as img:
                img.verify()
        except Exception:
            bad_files.append(path)

print("Bad/corrupt train images:", len(bad_files))

for f in bad_files[:20]:
    print(f)
```

```python
import tensorflow as tf
import os

bad_files = []

for root, _, files in os.walk("/content/classification_task"):
    for file in files:
        if file.lower().endswith((".jpg", ".jpeg", ".png")):
            path = os.path.join(root, file)
            try:
                data = tf.io.read_file(path)
                img = tf.image.decode_image(data, channels=3)
                _ = img.numpy()
            except Exception as e:
                bad_files.append((path, str(e)))

print("TensorFlow bad files:", len(bad_files))

for path, error in bad_files:
    print(path)
    print(error)
```

training


```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15,
    callbacks=[checkpoint]
)
```

best model


```python
from tensorflow.keras.models import load_model

best_model = load_model(
    "/content/drive/MyDrive/BRISC_ML_Project/features/efficientnetb7_best.keras"
)

print("Best EfficientNetB7 model loaded successfully!")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import numpy as np

y_true = []
y_pred = []

for images, labels in test_ds:
    predictions = best_model.predict(images, verbose=0)

    y_true.extend(labels.numpy())
    y_pred.extend(np.argmax(predictions, axis=1))

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average="macro")
recall = recall_score(y_true, y_pred, average="macro")
f1 = f1_score(y_true, y_pred, average="macro")

print("EfficientNet-B7 Final Test Results")
print("----------------------------------")
print(f"Accuracy        : {accuracy:.4f}")
print(f"Macro Precision : {precision:.4f}")
print(f"Macro Recall    : {recall:.4f}")
print(f"Macro F1        : {f1:.4f}")
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_true, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=test_ds.class_names
)

disp.plot(cmap="Blues", xticks_rotation=45)
plt.title("EfficientNet-B7 Confusion Matrix")
plt.tight_layout()

save_path = "/content/drive/MyDrive/BRISC_ML_Project/results/efficientnetb7_confusion_matrix.png"
plt.savefig(save_path, dpi=300, bbox_inches="tight")

plt.show()

print("Confusion matrix saved successfully! ✅")
print(save_path)
```
